```sage
sh(x)=(1/2)*(e^x - e^(-x))
ch(x)=(1/2)*(e^x + e^(-x))
th(x)=sh(x)/ch(x)
```

```sage
a=5
shp=plot(sh, -10,10, ymax=a, ymin=-a, legend_label="$\\mathrm{sh}(x)$")
chp=plot(ch, -10,10, ymax=a, ymin=-a, legend_label="$\\mathrm{ch}(x)$",color="red")
thp=plot(th, -10,10, ymax=a, ymin=-a, legend_label="$\\mathrm{th}(x)$",color="green")
shp+chp+thp
```
